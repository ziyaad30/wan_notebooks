```python
import os
from huggingface_hub import hf_hub_download

# 1. Define the correct absolute path for ComfyUI text encoders
target_dir = "ComfyUI/models/text_encoders"
os.makedirs(target_dir, exist_ok=True)

# 2. Download the Gemma text encoder
print("Downloading Gemma text encoder...")
hf_hub_download(
    repo_id="Comfy-Org/ltx-2",
    filename="split_files/text_encoders/gemma_3_12B_it_fp4_mixed.safetensors",
    local_dir=target_dir
)

# Move it out of the 'split_files/text_encoders/' subfolder into the main text_encoders folder
os.rename(
    os.path.join(target_dir, "split_files/text_encoders/gemma_3_12B_it_fp4_mixed.safetensors"),
    os.path.join(target_dir, "gemma_3_12B_it_fp4_mixed.safetensors")
)
# Clean up the empty nested folders left behind
os.removedirs(os.path.join(target_dir, "split_files/text_encoders"))

# 3. Download the LTX text projection model
print("Downloading LTX text projection...")
hf_hub_download(
    repo_id="Kijai/LTX2.3_comfy",
    filename="text_encoders/ltx-2.3_text_projection_bf16.safetensors",
    local_dir=target_dir
)

# Move it out of the 'text_encoders/' subfolder into the main folder
os.rename(
    os.path.join(target_dir, "text_encoders/ltx-2.3_text_projection_bf16.safetensors"),
    os.path.join(target_dir, "ltx-2.3_text_projection_bf16.safetensors")
)
# Clean up the empty nested folder left behind
os.rmdir(os.path.join(target_dir, "text_encoders"))

print("All text encoders successfully downloaded!")
```

```python
import os
from huggingface_hub import hf_hub_download

# Define paths
vae_dir = "ComfyUI/models/vae"
lora_dir = "ComfyUI/models/loras"

os.makedirs(vae_dir, exist_ok=True)
os.makedirs(lora_dir, exist_ok=True)

print("Downloading Sulphur Base LoRA...")
hf_hub_download(repo_id="SulphurAI/Sulphur-2-base", filename="sulphur_lora_rank_768.safetensors", local_dir=lora_dir)
# Rename to your custom filename: sulphur_final.safetensors
os.rename(os.path.join(lora_dir, "sulphur_lora_rank_768.safetensors"), os.path.join(lora_dir, "sulphur_final.safetensors"))

print("Downloading Distilled LoRA...")
hf_hub_download(repo_id="SulphurAI/Sulphur-2-base", filename="distill_loras/ltx-2.3-22b-distilled-lora-1.1_fro90_ceil72_condsafe.safetensors", local_dir=lora_dir)
os.rename(
    os.path.join(lora_dir, "distill_loras/ltx-2.3-22b-distilled-lora-1.1_fro90_ceil72_condsafe.safetensors"), 
    os.path.join(lora_dir, "ltx-2.3-22b-distilled-lora-1.1_fro90_ceil72_condsafe.safetensors")
)

# Clean up empty 'distill_loras' subfolder
os.rmdir(os.path.join(lora_dir, "distill_loras"))

print("All VAE and LoRA files successfully downloaded!")
```
