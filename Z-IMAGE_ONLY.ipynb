```python
import os
import time
import re
import sys

os.system("pkill -f 'ssh.*pinggy'")
time.sleep(2)

# Start tunnel and tee output to both file and stdout
os.system("ssh -p 443 -R0:localhost:8000 -o StrictHostKeyChecking=no -o ServerAliveInterval=30 UYekzW20QmT+force@a.pinggy.io 2>&1 | tee /workspace/tunnel.log &")

time.sleep(5)

with open("/workspace/tunnel.log") as f:
    for line in f:
        match = re.search(r'(https?://[^\s]+\.pinggy-free\.link)', line)
        if match:
            url = match.group(1)
            with open("/workspace/ngrok_url.txt", "w") as f:
                f.write(url)
            print(f"\nSaved URL: {url}")
            break
```
