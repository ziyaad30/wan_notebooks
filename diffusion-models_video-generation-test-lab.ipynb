```python
%%bash
# Kill existing ComfyUI process
if [ -f comfyui.pid ]; then
    PID=$(cat comfyui.pid)
    echo "Stopping ComfyUI (PID: $PID)..."
    kill $PID 2>/dev/null
    sleep 2
    
    # Force kill if still running
    if kill -0 $PID 2>/dev/null; then
        echo "Force killing..."
        kill -9 $PID
    fi
    rm comfyui.pid
fi

# Also kill any other ComfyUI processes
pkill -f "python3 main.py" 2>/dev/null
sleep 2

# Backup old log
if [ -f comfyui.log ]; then
    mv comfyui.log comfyui.log.old
fi

# Start ComfyUI with AMD optimization
APP_BASE=$(echo "$APP_URL" | sed 's|https://[^/]*||')

echo "Starting ComfyUI with AMD optimization..."
nohup python3 main.py \
  --listen 0.0.0.0 \
  --port 8000 \
  --enable-manager \
  --comfy-api-base "$APP_BASE" \
  > comfyui.log 2>&1 &

echo $! > comfyui.pid
echo "ComfyUI started with PID: $(cat comfyui.pid)"

# Wait and check
sleep 5
if kill -0 $(cat comfyui.pid) 2>/dev/null; then
    echo "✓ ComfyUI is running"
    echo ""
    echo "=== Recent logs ==="
    tail -15 comfyui.log
else
    echo "✗ ComfyUI failed to start. Check comfyui.log"
    tail -20 comfyui.log
fi
```
