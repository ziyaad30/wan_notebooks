```python
import subprocess

policy = """on_http_request:
  - expressions: []
    actions:
      - type: add-headers
        config:
          headers:
            ngrok-skip-browser-warning: "true"
"""

with open("/workspace/ngrok_policy.yml", "w") as f:
    f.write(policy)

print("✓ Policy file created")
```
